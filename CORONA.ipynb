```python
world_cases=[]
total_deaths=[]
mortality_rate=[]
total_recovered=[]
india_cases=[]

for i in dates:
  confirmed_sum = confirmed[i].sum()
  print(confirmed[i])
  death_sum=deaths[i].sum()
  recoverd_sum=recovered[i].sum()
  world_cases.append(int(confirmed_sum))
  total_deaths.append(int(death_sum))
  mortality_rate.append(int(death_sum/confirmed_sum))
  total_recovered.append(int(recoverd_sum))
  # india_cases.append(int(confirmed_cases[confirmed_cases['Country/Region']=='india'][i].sum()))
```

```python
# cheake each day cases in world

def eachday_increase(records):
  d=[]
  for i in range(len(records)):
    d.append(records[0])
  else:
    d.append(records[i]-records[i-1])
  return d
```

```python
world_daily_increase = eachday_increase(world_cases)
world_daily_increase
```

```python
model = LinearRegression()
model.fit(poly_x_train_confirmed,y_train_confirmed)
```

```python
test_pred = model.predict(poly_x_test_confirmed) # y_test_confirmed
```
